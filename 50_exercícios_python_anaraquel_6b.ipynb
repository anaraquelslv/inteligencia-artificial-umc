```python
# 2. Crie um programa que peça ao usuário dois números e exiba a soma deles.

'''Função que ler, valida se o usuário digitou um valor válido e soma ao valor anterior'''

def somar_numeros():
  '''Valor incremental para aparecer na saida quantos números foram digitados'''
  valordigitado = 1

  ''' Variável para armazenar a soma dos valores digitados'''
  soma = 0

  '''Loop infinito para ler os valores digitados pelo usuario'''

  while True:
    entrada = input(f"Digite o {valordigitado}° valor númerico para somar, ou digite \"f\" para parar: ")

    '''Se o usuario digitar "f", a função retorna a soma dos valores digitados'''
    if(entrada.lower() == "f"):
      '''tecnica early return'''
      return soma

    '''Se o usuário digitar um valor válido, a função soma o valor ao valor anterior'''
    try:
      numero = float(entrada)
      valordigitado += 1
      soma += numero

      '''Se o usuario digitar um valor inválido, a função exibe uma mensagem de erro e repete o loop'''
    except ValueError:
      print("Formato incorreto, digite novamente: ")



soma = somar_numeros()

print(f"Resultado da soma dos valores: {soma}")


# Alterei um pouco do que foi pedido originalmente para treinar idendação e criação de funções no python (estou mais acostumada com sintaxe do java)



```

```python
# 18 Crie um programa que verifique se um número fornecido é primo.

'''1. desconsiderei números menores ou iguais a um, inclui o numero 2 como primo (o unico numero par primo),
verifiquei que qualquer número maior que 2 e par não seja primo. Testesi todos os números impares a partir do 3 até a raiz do número
passado como argumento pulando de 2 em 2 (impares)
'''

def numero_primo(numero):
  if numero<= 1:
    return False
  if numero == 2:
    return True
  if numero > 2 and numero % 2 == 0:
    return False

  for i in range(3, int(numero**0.5) + 1, 2):
    if numero % i == 0:
      return False
  return True


entrada = input("Digite um numreo: ")

valor = int (entrada)
if numero_primo(valor):
  print("número primo")
else:
  print("número não primo")


```

```python
# 20. Dada uma matriz (lista de listas), crie sua matriz transposta.

'''
basicamente a gente coloca um marcador no começo e outro no fim da lista.
ai pega o valor  do meio e testa pra ver se é oq estamos procurando.
se o numero for maior que o do meio, a gente ignora a metade pra traz e foca so na da frente.
se for menor, foca na primeira metade.
vai dividindo assim ate achar o valor ou o espaço de busca acabar.
'''

def busca_binaria(lista, alvo):
    esquerda = 0
    direita = len(lista) - 1

    while esquerda <= direita:
        meio = (esquerda + direita) // 2

        if lista[meio] == alvo:
            return meio

        elif lista[meio] < alvo:
            esquerda = meio + 1

        else:
            direita = meio - 1

    return -1


numeros = [2, 3, 8, 12, 76, 23, 90, 56, 14, 91]
numeros.sort()
resultado = busca_binaria(numeros, 23)

print(numeros[resultado])
```

```python
# 22 Crie uma função para criptografar uma mensagem usando a cifra de César.

'''
a logica da cifra é ir passando por cada letra da frase e pular algumas posições no alfabeto.
pra fazer isso, converte a letra num numero, soma o pulo e dps converte pra letra de novo.
se passar do limite do alfabeto (passar do Z), a gente usa o resto da divisao por 26 pra voltar pro começo (pro A).
oq não for letra, tipo ponto ou espaço, a gente so repete igual tava e ta tudo certo.
'''

def cifra_cesar(mensagem, pulo):
    resultado = ""

    for letra in mensagem:
        # Verifica se é uma letra antes de tentar criptografar
        if letra.isalpha():
            # Define se é maiuscula ou minuscula para usar como base de cálculo
            base = ord('A') if letra.isupper() else ord('a')

            # Subtrai a base, soma o pulo, usa o % 26 pra não sair do alfabeto e soma a base de volta
            nova_letra = chr((ord(letra) - base + pulo) % 26 + base)
            resultado += nova_letra
        else:
            # Mantém espaços, números e pontuações do jeito que vieram
            resultado += letra

    return resultado

texto_original = "texto de teste"
texto_criptografado = cifra_cesar(texto_original, 3)

print("Original:", texto_original)
print("Criptografado:", texto_criptografado)
```

```python
# 23 Crie uma função para descriptografar uma mensagem cifrada com a cifra de César.


'''
pra descriptografar é basicamente a mesma logica, so que fazendo o caminho inverso.
em vez de somar o pulo pra ir pra frente no alfabeto, a gente subtrai esse valor pra voltar pras letras originais.
o python ajuda mt nisso pq o resto da divisao por 26 (% 26) ja lida sozinho caso a conta fique negativa (tipo voltar do A pro Z).
e igual antes, espaço e pontuação a gente so ignora e copia do mesmo jeito.
'''

def decifra_cesar(mensagem_cifrada, pulo):
    resultado = ""

    for letra in mensagem_cifrada:
        if letra.isalpha():
            base = ord('A') if letra.isupper() else ord('a')

            # A única mudança matemática em relação à cifra é o sinal de menos (-) antes do pulo
            letra_original = chr((ord(letra) - base - pulo) % 26 + base)
            resultado += letra_original
        else:
            resultado += letra

    return resultado


texto_criptografado = "whawr gh whvwh"
texto_revelado = decifra_cesar(texto_criptografado, 3)

print("Criptografado:", texto_criptografado)
print("Descriptografado:", texto_revelado)
```

```python
# 23 Implemente uma função para validar um CPF brasileiro.

'''
a logica de validar cpf é primeiro limpar a string tirando ponto e traço pra ficar so os 11 numeros.
se não tiver 11 numeros ou for tudo repetido tipo 111.111.111-11 a gente ja sabe q é falso e recusa.
dps a gente calcula os dois ultimos digitos verificadores usando a regra da receita federal.
tem q ir multiplicando cada numero por uns pesos (de 10 caindo pra 2) e ver o resto da divisao por 11.
calcula o primeiro digito e ve se bate, dps calcula o segundo e faz a msm coisa. se bater tudo é valido.
'''

import re

def valida_cpf(cpf):
    # Pega apenas os números digitados, ignorando pontos e traços
    numeros = re.sub(r'[^0-9]', '', str(cpf))

    # Verifica tamanho ou se é uma sequência de números iguais (ex: 00000000000)
    if len(numeros) != 11 or numeros == numeros[0] * 11:
        return False

    # Calcula o primeiro dígito verificador
    soma_1 = 0
    for i in range(9):
        soma_1 += int(numeros[i]) * (10 - i)

    digito_1 = (soma_1 * 10) % 11
    if digito_1 == 10:
        digito_1 = 0

    if digito_1 != int(numeros[9]):
        return False

    # Calcula o segundo dígito verificador
    soma_2 = 0
    for i in range(10):
        soma_2 += int(numeros[i]) * (11 - i)

    digito_2 = (soma_2 * 10) % 11
    if digito_2 == 10:
        digito_2 = 0

    if digito_2 != int(numeros[10]):
        return False

    return True

# Testando o código
cpf_teste = "123.456.789-00" # CPF inválido qualquer de exemplo
resultado = valida_cpf(cpf_teste)

print(f"O CPF {cpf_teste} é válido? {resultado}")
```

```python
# 24 Classifique uma lista de números em três categorias: positivos, negativos e zeros.

'''criei um dicionário com três listas vazias no começo.
Depois, fiz um loop para ler cada número da lista original.
Se o número for maior que zero, ele vai para a lista de positivos; se for menor, entra nos negativos.
O que sobrar, cai  na lista de zeros.
'''

def classifica_numeros(lista):
    resultado = {
        'positivos': [],
        'negativos': [],
        'zeros': []
    }

    for numero in lista:
        if numero > 0:
            resultado['positivos'].append(numero)
        elif numero < 0:
            resultado['negativos'].append(numero)
        else:
            resultado['zeros'].append(numero)

    return resultado

# Testando o código
lista_misturada = [15, -4, 0, 42, -9, 0, 3, -1]

# Agora guardamos tudo em uma variável só
categorias = classifica_numeros(lista_misturada)

print("Original:", lista_misturada)
print("Positivos:", categorias['positivos'])
print("Negativos:", categorias['negativos'])
print("Zeros:", categorias['zeros'])
```
