```python
# 20. Dada uma matriz (lista de listas), crie sua matriz transposta.

'''
basicamente a gente coloca um marcador no começo e outro no fim da lista.
ai pega o valor  do meio e testa pra ver se é oq estamos procurando.
se o numero for maior que o do meio, a gente ignora a metade pra traz e foca so na da frente.
se for menor, foca na primeira metade.
vai dividindo assim ate achar o valor ou o espaço de busca acabar.
'''

def busca_binaria(lista, alvo):
    esquerda = 0
    direita = len(lista) - 1

    while esquerda <= direita:
        meio = (esquerda + direita) // 2

        if lista[meio] == alvo:
            return meio

        elif lista[meio] < alvo:
            esquerda = meio + 1

        else:
            direita = meio - 1

    return -1


numeros = [2, 3, 8, 12, 76, 23, 90, 56, 14, 91]
numeros.sort()
resultado = busca_binaria(numeros, 23)

print(numeros[resultado])
```

```python
# 22 Crie uma função para criptografar uma mensagem usando a cifra de César.

'''
a logica da cifra é ir passando por cada letra da frase e pular algumas posições no alfabeto.
pra fazer isso, converte a letra num numero, soma o pulo e dps converte pra letra de novo.
se passar do limite do alfabeto (passar do Z), a gente usa o resto da divisao por 26 pra voltar pro começo (pro A).
oq não for letra, tipo ponto ou espaço, a gente so repete igual tava e ta tudo certo.
'''

def cifra_cesar(mensagem, pulo):
    resultado = ""

    for letra in mensagem:
        # Verifica se é uma letra antes de tentar criptografar
        if letra.isalpha():
            # Define se é maiuscula ou minuscula para usar como base de cálculo
            base = ord('A') if letra.isupper() else ord('a')

            # Subtrai a base, soma o pulo, usa o % 26 pra não sair do alfabeto e soma a base de volta
            nova_letra = chr((ord(letra) - base + pulo) % 26 + base)
            resultado += nova_letra
        else:
            # Mantém espaços, números e pontuações do jeito que vieram
            resultado += letra

    return resultado

texto_original = "texto de teste"
texto_criptografado = cifra_cesar(texto_original, 3)

print("Original:", texto_original)
print("Criptografado:", texto_criptografado)
```

```python
# 23 Crie uma função para descriptografar uma mensagem cifrada com a cifra de César.


'''
pra descriptografar é basicamente a mesma logica, so que fazendo o caminho inverso.
em vez de somar o pulo pra ir pra frente no alfabeto, a gente subtrai esse valor pra voltar pras letras originais.
o python ajuda mt nisso pq o resto da divisao por 26 (% 26) ja lida sozinho caso a conta fique negativa (tipo voltar do A pro Z).
e igual antes, espaço e pontuação a gente so ignora e copia do mesmo jeito.
'''

def decifra_cesar(mensagem_cifrada, pulo):
    resultado = ""

    for letra in mensagem_cifrada:
        if letra.isalpha():
            base = ord('A') if letra.isupper() else ord('a')

            # A única mudança matemática em relação à cifra é o sinal de menos (-) antes do pulo
            letra_original = chr((ord(letra) - base - pulo) % 26 + base)
            resultado += letra_original
        else:
            resultado += letra

    return resultado


texto_criptografado = "whawr gh whvwh"
texto_revelado = decifra_cesar(texto_criptografado, 3)

print("Criptografado:", texto_criptografado)
print("Descriptografado:", texto_revelado)
```

```python
# 24 Implemente uma função para validar um CPF brasileiro.

'''
a logica de validar cpf é primeiro limpar a string tirando ponto e traço pra ficar so os 11 numeros.
se não tiver 11 numeros ou for tudo repetido tipo 111.111.111-11 a gente ja sabe q é falso e recusa.
dps a gente calcula os dois ultimos digitos verificadores usando a regra da receita federal.
tem q ir multiplicando cada numero por uns pesos (de 10 caindo pra 2) e ver o resto da divisao por 11.
calcula o primeiro digito e ve se bate, dps calcula o segundo e faz a msm coisa. se bater tudo é valido.
'''

import re

def valida_cpf(cpf):
    # Pega apenas os números digitados, ignorando pontos e traços
    numeros = re.sub(r'[^0-9]', '', str(cpf))

    # Verifica tamanho ou se é uma sequência de números iguais (ex: 00000000000)
    if len(numeros) != 11 or numeros == numeros[0] * 11:
        return False

    # Calcula o primeiro dígito verificador
    soma_1 = 0
    for i in range(9):
        soma_1 += int(numeros[i]) * (10 - i)

    digito_1 = (soma_1 * 10) % 11
    if digito_1 == 10:
        digito_1 = 0

    if digito_1 != int(numeros[9]):
        return False

    # Calcula o segundo dígito verificador
    soma_2 = 0
    for i in range(10):
        soma_2 += int(numeros[i]) * (11 - i)

    digito_2 = (soma_2 * 10) % 11
    if digito_2 == 10:
        digito_2 = 0

    if digito_2 != int(numeros[10]):
        return False

    return True

# Testando o código
cpf_teste = "123.456.789-00" # CPF inválido qualquer de exemplo
resultado = valida_cpf(cpf_teste)

print(f"O CPF {cpf_teste} é válido? {resultado}")
```

```python
# 25 Classifique uma lista de números em três categorias: positivos, negativos e zeros.

'''criei um dicionário com três listas vazias no começo.
Depois, fiz um loop para ler cada número da lista original.
Se o número for maior que zero, ele vai para a lista de positivos; se for menor, entra nos negativos.
O que sobrar, cai  na lista de zeros.
'''

def classifica_numeros(lista):
    resultado = {
        'positivos': [],
        'negativos': [],
        'zeros': []
    }

    for numero in lista:
        if numero > 0:
            resultado['positivos'].append(numero)
        elif numero < 0:
            resultado['negativos'].append(numero)
        else:
            resultado['zeros'].append(numero)

    return resultado

# Testando o código
lista_misturada = [15, -4, 0, 42, -9, 0, 3, -1]

# Agora guardamos tudo em uma variável só
categorias = classifica_numeros(lista_misturada)

print("Original:", lista_misturada)
print("Positivos:", categorias['positivos'])
print("Negativos:", categorias['negativos'])
print("Zeros:", categorias['zeros'])
```

```python
# 26 Crie uma calculadora que possa realizar adição, subtração, multiplicação e divisão.

'''
criei uma função separada pra fazer a conta msm e usar o early return que eu acostumei.
o operador recebido dita qual conta fazer. e tbm coloquei um teste rapido na divisao pra
não deixar o programa estourar erro se tentar dividir por zero.
dps usei o while True de sempre pra ler os inputs e o try except pra garantir q sao numeros validos
antes de mandar pra função.
'''
def calcular_valores(num1, num2, operador):
  if operador == '+':
    return num1 + num2
  if operador == '-':
    return num1 - num2
  if operador == '*':
    return num1 * num2
  if operador == '/':
    if num2 == 0:
      return "Não da pra dividir por zero"
    return num1 / num2

  return "Operador não reconhecido"

while True:
  print("\n--- Calculadora (digite 'f' pra fechar) ---")
  entrada1 = input("Digite o 1° número: ")

  if entrada1.lower() == 'f':
    break

  operacao = input("Digite a operação (+, -, *, /): ")
  entrada2 = input("Digite o 2° número: ")

  try:
    valor1 = float(entrada1)
    valor2 = float(entrada2)

    resultadoDaConta = calcular_valores(valor1, valor2, operacao)
    print(f"Resultado da operação: {resultadoDaConta}")

  except ValueError:
    print("Valor incorreto, digite novamente")
```

```python
# 27 Verifique se duas palavras são anagramas (ou seja, se são formadas pelas mesmas letras).

'''
pra saber se e anagrama a logica mais facil e tirar os espacos em branco,
deixar tudo minusculo e dps ordenar as letras em ordem alfabetica.
se a gente ordenar as duas palavras e as listas ficarem iguais, entao elas tem as msm
letras e sao anagramas. a funcao sorted do python ja faz isso.


'''
def e_anagrama(palavra1, palavra2):
  str1 = palavra1.replace(" ", "").lower()
  str2 = palavra2.replace(" ", "").lower()

  if len(str1) != len(str2):
    return False

  if sorted(str1) == sorted(str2):
    return True

  return False


while True:
  print("\n--- Teste de Anagrama ---")
  palavraUm = input("Digite a 1° palavra ou 'f' pra sair: ")

  if palavraUm.lower() == "f":
    break

  palavraDois = input("Digite a 2° palavra: ")

  if e_anagrama(palavraUm, palavraDois):
    print(f"Sim! \"{palavraUm}\" e \"{palavraDois}\" são anagramas")
  else:
    print(f"Não, \"{palavraUm}\" e \"{palavraDois}\" NÃO são anagramas")
```

```python
# 30 Desenvolva um jogo simples da forca onde o usuário deve adivinhar uma palavra.

'''
pra fazer a forca eu importei o modulo random pra escolher uma palavra aleatoria de uma lista.
criei uma variavel pra guardar as letras q o usuario ja digitou e uma pra contar quantas chances ele tem.
dentro do loop a logica principal e montar a palavra secreta iterando sob as letras da palavra escolhida.
se a letra ta na lista de acertadas, mostra ela, senao mostra um underline "_".
dps confere se ainda tem underline, se nao tiver o loop para pq ele ganhou.
valido tbm se o usuario digitou uma letra valida usando o isalpha() ou se ja digitou ela antes pra nao perder chance atoa.
coloquei tudo dentro de uma funcao pra usar o early return quando o jogo acaba.
'''
import random

def jogo_forca():
  palavras = ["python", "java", "desenvolvimento", "computador", "teclado"]
  palavraEscolhida = random.choice(palavras).upper()
  letrasTentadas = []
  chances = 6

  print("--- Jogo da Forca ---")

  while chances > 0:
    '''monta a palavra com os acertos'''
    palavraEscondida = ""
    for letra in palavraEscolhida:
      if letra in letrasTentadas:
        palavraEscondida += letra + " "
      else:
        palavraEscondida += "_ "

    print(f"\nPalavra: {palavraEscondida}")
    print(f"Chances restantes: {chances}")

    # formata a lista de letras tentadas separada por virgula pra ficar mais visual
    textoLetras = ", ".join(letrasTentadas)
    print(f"Letras ja tentadas: {textoLetras}")

    '''early return se nao tem mais underline, ou seja, acertou tudo'''
    if "_" not in palavraEscondida:
      print(f"\nParabens! Vc adivinhou a palavra: {palavraEscolhida}")
      return

    entrada = input("Digite uma letra: ").upper()

    '''valida se e apenas uma letra msm'''
    if len(entrada) != 1 or not entrada.isalpha():
      print("Formato incorreto, digite apenas UMA letra.")
      continue

    if entrada in letrasTentadas:
      print("Vc ja tentou essa letra. Tente outra.")
      continue

    letrasTentadas.append(entrada)

    if entrada in palavraEscolhida:
      print("Boa! Letra correta.")
    else:
      print("Errou! A letra nao ta na palavra.")
      chances -= 1

  '''se o while acabar e pq as chances chegaram em 0'''
  print(f"\nGame Over! A palavra era {palavraEscolhida}")

jogo_forca()
```

```python
# 31 Implemente o algoritmo de ordenação por bolha (Bubble Sort).

'''
a logica do bubble sort (ordenacao por bolha) e ir iterando sob a lista e comparando os numeros de 2 em 2.
se o numero da esquerda for maior q o da direita, a gente inverte eles de lugar usando uma variavel temp.
dps a gente repete isso ate chegar no final. o maior numero vai "borbulhar" pro final da lista.
ai o loop de fora garante q a gente vai fazer isso pra todos os elementos.
coloquei uma flag pra ver se teve troca, se rodar o laco de dentro e nao trocar nada,
ja da pra usar o break e parar pq a lista ja ta ordenada e a gente economiza processamento.
'''
def bubble_sort(listaNumeros):
  tamanhoLista = len(listaNumeros)

  for i in range(tamanhoLista):
    trocou = False

    '''o -i-1 e pq os ultimos numeros ja vao estar ordenados no final, entao nao precisa checar eles de novo'''
    for j in range(0, tamanhoLista - i - 1):
      if listaNumeros[j] > listaNumeros[j + 1]:
        '''usa a variavel temporaria pra nao perder o valor na hora de inverter os dois'''
        temp = listaNumeros[j]
        listaNumeros[j] = listaNumeros[j + 1]
        listaNumeros[j + 1] = temp
        trocou = True

    '''se iterar sob a lista inteira e a variavel continuar false, quebra o laco de repeticao'''
    if not trocou:
      break

  return listaNumeros

numeros = [64, 34, 25, 12, 22, 11, 90]
print(f"Lista original: {numeros}")

listaOrdenada = bubble_sort(numeros)
print(f"Lista ordenada: {listaOrdenada}")

```

```python
# 32 Crie um jogo da velha para dois jogadores.

'''
pra fazer o jogo da velha eu usei uma lista com 9 posicoes, iniciando com os numeros de 1 a 9
pra facilitar pro usuario escolher onde jogar.
criei funcoes separadas pra imprimir o tabuleiro, pra ver se alguem ganhou e o loop principal do jogo.
no loop do jogo eu fico alternando a vez do jogadorAtual entre X e O.
coloquei um try except pra validar se o usuario digitou um numero msm e um if pra ver se a posicao ja nao ta ocupada.
se alguem ganhar ou der velha (empate), eu uso o return (early return de novo) pra sair do laco de repeticao e finalizar.
'''

def imprimir_tabuleiro(tabuleiro):
  print(f"\n {tabuleiro[0]} | {tabuleiro[1]} | {tabuleiro[2]} ")
  print("---|---|---")
  print(f" {tabuleiro[3]} | {tabuleiro[4]} | {tabuleiro[5]} ")
  print("---|---|---")
  print(f" {tabuleiro[6]} | {tabuleiro[7]} | {tabuleiro[8]} \n")

def verificar_vencedor(tabuleiro, jogador):
  '''cria uma lista de listas com todas as combinacoes de vitoria (linhas, colunas e diagonais)'''
  vitorias = [
      [0, 1, 2], [3, 4, 5], [6, 7, 8],
      [0, 3, 6], [1, 4, 7], [2, 5, 8],
      [0, 4, 8], [2, 4, 6]
  ]

  '''itera sob as condicoes pra ver se as 3 posicoes tem a string do jogador atual'''
  for condicao in vitorias:
    if tabuleiro[condicao[0]] == tabuleiro[condicao[1]] == tabuleiro[condicao[2]] == jogador:
      return True
  return False

def jogo_da_velha():
  tabuleiro = ['1', '2', '3', '4', '5', '6', '7', '8', '9']
  jogadorAtual = 'X'
  jogadas = 0

  print("--- Jogo da Velha ---")

  while True:
    imprimir_tabuleiro(tabuleiro)
    entrada = input(f"Vez do jogador {jogadorAtual}. Escolha uma posicao (1-9): ")

    try:
      '''subtrai 1 pq o indice da lista comeca no 0'''
      posicao = int(entrada) - 1

      if posicao < 0 or posicao > 8:
        print("Posicao invalida, digite um numero entre 1 e 9")
        continue

      if tabuleiro[posicao] == 'X' or tabuleiro[posicao] == 'O':
        print("Essa posicao ja ta ocupada. Escolha outra.")
        continue

      '''atualiza o tabuleiro com o X ou O'''
      tabuleiro[posicao] = jogadorAtual
      jogadas += 1

      '''verifica se o jogador que acabou de jogar ganhou'''
      if verificar_vencedor(tabuleiro, jogadorAtual):
        imprimir_tabuleiro(tabuleiro)
        print(f"Parabens! O jogador {jogadorAtual} venceu!")
        return

      '''se preencheu as 9 posicoes e ninguem ganhou, avisa q empatou'''
      if jogadas == 9:
        imprimir_tabuleiro(tabuleiro)
        print("Deu velha! Empate.")
        return

      '''troca a vez do jogador pro proximo loop'''
      if jogadorAtual == 'X':
        jogadorAtual = 'O'
      else:
        jogadorAtual = 'X'

    except ValueError:
      print("Formato incorreto, digite apenas numeros de 1 a 9.")

jogo_da_velha()
```

```python
# 33 Verifique se duas palavras fornecidas são anagramas uma da outra.

'''
como a gente ja resolveu um exercicio de anagrama antes usando a funcao sorted(),
agr eu decidi fazer usando a msm logica do exercicio de contar letras (exercicio 12)
pra treinar a manipulacao de dicionarios e nao ficar dependente so de funcao pronta.

criei um dicionario pra guardar a frequencia de cada letra da primeira palavra.
dps eu itero sob a segunda palavra e vou subtraindo a contagem do dicionario.
se no final sobrar algum valor diferente de 0 no dicionario, e pq nao tem as msm letras
e nao e anagrama. tbm uso o early return no comeco se os tamanhos forem diferentes pra
economizar processamento logo de cara.
'''
def e_anagrama_manual(palavra1, palavra2):
  str1 = palavra1.replace(" ", "").lower()
  str2 = palavra2.replace(" ", "").lower()

  if len(str1) != len(str2):
    return False

  frequencia = {}

  '''conta as letras da primeira palavra'''
  for letra in str1:
    if letra in frequencia:
      frequencia[letra] += 1
    else:
      frequencia[letra] = 1

  '''desconta com as letras da segunda palavra'''
  for letra in str2:
    if letra in frequencia:
      frequencia[letra] -= 1
    else:
      '''se a letra nem existe no dicionario, ja sabe q nao e anagrama'''
      return False

  '''valida se o contador de todas as letras zerou'''
  for qtd in frequencia.values():
    if qtd != 0:
      return False

  return True

while True:
  print("\n--- Teste de Anagrama (Logica Manual) ---")
  palavraUm = input("Digite a 1° palavra ou 'f' pra sair: ")

  if palavraUm.lower() == "f":
    break

  palavraDois = input("Digite a 2° palavra: ")

  if e_anagrama_manual(palavraUm, palavraDois):
    print(f"Sim! \"{palavraUm}\" e \"{palavraDois}\" são anagramas")
  else:
    print(f"Não, \"{palavraUm}\" e \"{palavraDois}\" NÃO são anagramas")
```

```python
# 35 Gere a sequência de Fibonacci até um número n fornecido pelo usuário.

'''
como a gente ja fez fibonacci no exercicio 10, a logica central e a msm (usando a variavel temp).
a diferenca e q em vez de rodar um numero fixo de vezes com o for, eu usei o while pra ir
gerando a sequencia ate o valor atual (a nossa variavel b) passar do numero limite q o usuario passou.
criei a funcao separada com a logica principal e dps aquele msm esquema do while True
e try except pra validar a entrada de dados.
'''
def gerar_fibonacci(limite):
  a = 0
  b = 1

  '''imprime o zero primeiro pra começar a sequencia'''
  print(a)

  '''enquanto o proximo numero da sequencia for menor ou igual ao limite, ele continua gerando'''
  while b <= limite:
    print(b)
    temp = a
    a = b
    b = temp + b

while True:
  print("\n--- Gerador de Fibonacci ---")
  entrada = input("Digite ate qual numero maximo vc quer gerar a sequencia (ou 'f' pra sair): ")

  if entrada.lower() == 'f':
    break

  try:
    numeroLimite = int(entrada)

    if numeroLimite < 0:
      print("O valor nao pode ser negativo. Tente de novo.")
      continue

    print(f"\nSequencia ate o numero {numeroLimite}:")
    gerar_fibonacci(numeroLimite)

  except ValueError:
    print("Formato incorreto, vc precisa digitar um numero inteiro valido.")
```

```python
# 36 Crie um gerador de senhas aleatórias com letras e números.

'''
pra fazer o gerador eu importei a biblioteca random pra sortear e a biblioteca string
q ja tem umas constantes com todas as letras e numeros prontos, ai nao preciso digitar tudo na mao.

a logica da funcao e juntar as letras (maiusculas e minusculas) e os numeros numa base so.
dps criei um laco for q vai rodar a quantidade de vezes q o usuario pediu (o tamanho da senha)
e vai usando o random.choice pra iterar sob a base e escolher um caractere aleatorio pra ir
juntando no resultado final.

no input eu mantive o nosso padrao do while True com try except pra garantir q o tamanho da
senha seja msm um numero inteiro valido antes de chamar a funcao.
'''
import random
import string

def gerar_senha(tamanho):
  '''junta ascii_letters (todas as letras) e digits (numeros de 0 a 9)'''
  caracteresBase = string.ascii_letters + string.digits
  senha = ""

  for i in range(tamanho):
    senha += random.choice(caracteresBase)

  return senha

while True:
  print("\n--- Gerador de Senhas ---")
  entrada = input("Digite o tamanho da senha q vc quer (ou 'f' pra sair): ")

  if entrada.lower() == 'f':
    break

  try:
    tamanhoSenha = int(entrada)

    if tamanhoSenha <= 0:
      print("O tamanho da senha precisa ser maior q zero. Tente de novo.")
      continue

    senhaCriada = gerar_senha(tamanhoSenha)
    print(f"Sua nova senha gerada: {senhaCriada}")

  except ValueError:
    print("Formato incorreto, digite um numero inteiro valido.")
```

```python
# 37 Implemente uma função recursiva para calcular o fatorial de um número.

'''
como a gente ja fez fatorial no exercicio 9 usando o laco for, agr a ideia e usar recursao.
a logica da recursao e a funcao chamar ela msm ate chegar num caso base pra parar, senao ela fica num loop infinito e estoura a memoria.
o caso base do fatorial e o 0 ou 1, q sempre vao retornar 1. uso o early return pra isso ja no comeco.
se o numero for maior q 1, a funcao pega o numero atual e multiplica pelo resultado dela msm chamando o numero anterior.
dps coloquei aquele nosso msm padrao do while True com try except pra garantir q o usuario digite um numero inteiro positivo valido.
'''
def calcular_fatorial_recursivo(numero):
  if numero == 0 or numero == 1:
    return 1

  return numero * calcular_fatorial_recursivo(numero - 1)

while True:
  print("\n--- Fatorial Recursivo ---")
  entrada = input("Digite um numero inteiro positivo (ou 'f' pra sair): ")

  if entrada.lower() == 'f':
    break

  try:
    valor = int(entrada)

    if valor < 0:
      print("O valor nao pode ser negativo. Tente de novo.")
      continue

    resultadoFatorial = calcular_fatorial_recursivo(valor)
    print(f"O fatorial de {valor} é {resultadoFatorial}")

  except ValueError:
    print("Valor incorreto, vc precisa digitar um numero inteiro valido.")

```

```python
# 38 Conte o número de palavras em um texto fornecido pelo usuário.

'''
como a gente ja fez algo parecido antes pra contar frequencia, contar palavras e bem mais de boa.
a logica e usar a funcao split() nativa do python q divide o texto toda vez q acha um espaco em branco e transforma numa lista.
dps disso e so usar a funcao len() pra ver o tamanho da lista q o split() gerou e a gente tem a quantidade de palavras.
coloquei uma validacao simples usando o early return na funcao caso o usuario de enter sem digitar nada ou so digite espaco.
e mantive o while True pra iterar sob varios testes sem precisar rodar a celula do colab toda hora.
'''
def contar_palavras(texto):
  '''se a string tiver vazia ou for so espaco, retorna 0 direto (early return)'''
  if not texto.strip():
    return 0

  listaPalavras = texto.split()
  return len(listaPalavras)

while True:
  print("\n--- Contador de Palavras ---")
  entrada = input("Digite um texto ou frase (ou 'f' pra sair): ")

  if entrada.lower() == 'f':
    break

  qtdPalavras = contar_palavras(entrada)
  print(f"O texto q vc digitou tem {qtdPalavras} palavras.")
```

```python
# 40 Crie um programa para gerenciar uma lista de compras, permitindo adicionar, remover e listar itens.

'''
criei uma lista vazia pra ir guardando os itens.
fiz funcoes separadas pra cada acao (adicionar, remover, listar) pra aplicar o principio da responsabilidade unica.
no remover eu valido primeiro se o item existe na lista usando o 'in', senao o python estoura erro se tentar remover oq nao existe.
dps montei um menu simples dentro do nosso while True padrao pra iterar sob as opcoes do usuario e ir manipulando a lista.
'''
listaDeCompras = []

def adicionar_item(item):
  listaDeCompras.append(item)
  print(f"'{item}' foi adicionado na lista.")

def remover_item(item):
  if item in listaDeCompras:
    listaDeCompras.remove(item)
    print(f"'{item}' foi removido da lista.")
  else:
    print(f"O item '{item}' nao ta na lista, entao nao da pra remover.")

def listar_itens():
  if len(listaDeCompras) == 0:
    print("A lista de compras ta vazia.")
    '''early return pra nao precisar do else'''
    return

  print("\n--- Sua Lista de Compras ---")
  '''itera sob os itens da lista pra printar um por um com um traco na frente pra ficar bonito'''
  for item in listaDeCompras:
    print(f"- {item}")
  print("----------------------------\n")

while True:
  print("\n--- Gerenciador de Compras ---")
  print("1 - Adicionar item")
  print("2 - Remover item")
  print("3 - Listar itens")
  print("4 - Sair (ou digite 'f')")

  opcao = input("Escolha oq vc quer fazer: ")

  if opcao == '4' or opcao.lower() == 'f':
    print("Saindo do gerenciador...")
    break

  if opcao == '1':
    novoItem = input("Digite o nome do item pra adicionar: ")
    adicionar_item(novoItem)

  elif opcao == '2':
    itemPraRemover = input("Digite o nome do item pra remover: ")
    remover_item(itemPraRemover)

  elif opcao == '3':
    listar_itens()

  else:
    print("Opcao incorreta, digite um numero do menu.")

```

```python
# 41 Remova duplicatas de uma lista de números.

'''
tem duas formas de fazer isso. a forma mais facil no python e usar a funcao set() q ja
tira todas as duplicatas nativamente, mas ela nao garante a ordem dos numeros.
entao eu resolvi fazer a logica na mao pra manter a ordem original dos dados.

criei uma lista vazia chamada listaLimpa e vou iterando sob a lista q a gente recebe da funcao.
se o numero atual ainda nao existir na lista nova, a gente adiciona usando o append. se ja existir, o for ignora.
dps no while de input usei o split() pra ler os numeros digitados tudo de uma vez separado por espaco.
'''
def tirar_duplicatas(listaComRepeticao):
  listaLimpa = []

  for numero in listaComRepeticao:
    if numero not in listaLimpa:
      listaLimpa.append(numero)

  return listaLimpa

while True:
  print("\n--- Removedor de Duplicatas ---")
  entrada = input("Digite os numeros separados por espaco (ou 'f' pra sair): ")

  if entrada.lower() == 'f':
    break

  try:
    '''
    usei list comprehension pra fazer o for direto na criacao da lista e
    ja transformar oq vem do split em float pra facilitar
    '''
    listaNumeros = [float(x) for x in entrada.split()]

    listaResultado = tirar_duplicatas(listaNumeros)

    print(f"Lista original: {listaNumeros}")
    print(f"Lista sem as duplicatas: {listaResultado}")

  except ValueError:
    print("Formato incorreto, certifique q digitou apenas numeros separados por espaco.")
```

```python
# 42 Gere todos os números primos até um número n fornecido.

'''
como a gente ja fez uma funcao pra testar se o numero e primo no exercicio 18,
eu vou reaproveitar a logica dela aqui pra deixar as coisas mais organizadas.
criei uma funcao nova q cria uma lista vazia e vai iterando sob os numeros de 2 ate o limite q o usuario digitou.
se o numero atual passar no teste da funcao e_primo, a gente adiciona ele na lista usando o append.
dps mantive o nosso while True padrao com try except pra validar se o usuario digitou um limite inteiro valido.
'''
def e_primo(numero):
  if numero <= 1:
    return False
  if numero == 2:
    return True
  if numero > 2 and numero % 2 == 0:
    return False

  for i in range(3, int(numero**0.5) + 1, 2):
    if numero % i == 0:
      return False

  return True

def gerar_primos(limite):
  listaPrimos = []

  for i in range(2, limite + 1):
    if e_primo(i):
      listaPrimos.append(i)

  return listaPrimos

while True:
  print("\n--- Gerador de Numeros Primos ---")
  entrada = input("Digite ate qual numero vc quer achar os primos (ou 'f' pra sair): ")

  if entrada.lower() == 'f':
    break

  try:
    limiteMaximo = int(entrada)

    if limiteMaximo < 2:
      print("Nao tem numero primo antes do 2. Digite um limite maior q 1.")
      continue

    resultadoLista = gerar_primos(limiteMaximo)
    print(f"Numeros primos ate {limiteMaximo}:\n{resultadoLista}")

  except ValueError:
    print("Formato incorreto, vc precisa digitar um numero inteiro valido.")
```

```python
'''
pra fazer o conversor eu usei um dicionario com os fatores de conversao usando metros como base.
assim fica mt mais facil pq eu transformo qualquer medida q o usuario digitar pra metros primeiro
e dps divido pelo fator da medida q ele quer como resultado. isso me poupa de fazer um monte de if e else.
criei a funcao com early return caso a unidade q a pessoa digitou nao exista no nosso dicionario.
dps montei aquele loop while True padrao pra ler o valor, a unidade de origem e a de destino com o try except.
'''
def converter_comprimento(valor, origem, destino):
  '''dicionario usando metro como base (1)'''
  fatores = {
      'km': 1000.0,
      'm': 1.0,
      'cm': 0.01,
      'mm': 0.001
  }

  '''early return se a unidade for invalida'''
  if origem not in fatores or destino not in fatores:
    return None

  '''transforma tudo pra metros primeiro e dps pro destino'''
  valorEmMetros = valor * fatores[origem]
  resultado = valorEmMetros / fatores[destino]

  return resultado

while True:
  print("\n--- Conversor de Medidas ---")
  print("Unidades disponiveis: km, m, cm, mm")

  entrada = input("Digite o valor q vc quer converter (ou 'f' pra sair): ")

  if entrada.lower() == 'f':
    break

  try:
    valorParaConverter = float(entrada)

    unidadeOrigem = input("De qual unidade? (ex: m): ").lower()
    unidadeDestino = input("Para qual unidade? (ex: cm): ").lower()

    resultadoConversao = converter_comprimento(valorParaConverter, unidadeOrigem, unidadeDestino)

    if resultadoConversao is None:
      print("Unidade invalida. Da uma olhada nas unidades disponiveis e tenta de novo.")
    else:
      print(f"\nResultado: {valorParaConverter}{unidadeOrigem} = {resultadoConversao}{unidadeDestino}")

  except ValueError:
    print("Formato incorreto, digite apenas numeros pro valor.")
```

```python
# 44 Gere a sequência de Fibonacci até um determinado número de termos.
'''
criei uma lista vazia e usei um laco for com o range rodando a quantidade exata de vezes q o usuario pediu.
em cada iteracao a gente adiciona o 'a' na lista e dps usa aquela msm variavel temp de antes pra i
 atualizando os valores do a e do b.
mantive o while True com try except pra validar a entrada de dados e usei o early return
 caso a pessoa peça 0 termos ou numero negativo.
'''
def fibonacci_termos(qtdTermos):
  if qtdTermos <= 0:
    return []

  listaFibo = []
  a = 0
  b = 1

  for i in range(qtdTermos):
    listaFibo.append(a)
    temp = a
    a = b
    b = temp + b

  return listaFibo

while True:
  print("\n--- Fibonacci por Termos ---")
  entrada = input("Digite quantos termos da sequencia vc quer ver (ou 'f' pra sair): ")

  if entrada.lower() == 'f':
    break

  try:
    quantidade = int(entrada)

    if quantidade <= 0:
      print("A quantidade de termos precisa ser maior q zero. Tente de novo.")
      continue

    resultadoFibo = fibonacci_termos(quantidade)
    print(f"Os primeiros {quantidade} termos da sequencia:\n{resultadoFibo}")

  except ValueError:
    print("Formato incorreto, vc precisa digitar um numero inteiro valido.")
```

```python
# 45 Classifique os números de uma lista como pares ou ímpares.

'''
criei uma funcao q recebe a lista de numeros e devolve um dicionario.
nesse dicionario eu coloquei duas listas separadas, uma pros pares e outra pros impares.
ai uso um laco for pra iterar sob os numeros e dps o operador de modulo pra saber se o resto da divisao por 2 e zero (par) ou nao (impar).
no input do while a gente pega a string, usa o split e ja converte tudo pra int usando list comprehension q nem fizemos no exercicio de tirar duplicatas.
coloquei tbm um if rapidinho ali pra testar se a pessoa so deu enter sem digitar nenhum numero.
'''

def classificar_pares_impares(listaNumeros):
  resultado = {
    'pares': [],
    'impares': []
  }

  for numero in listaNumeros:
    if numero % 2 == 0:
      resultado['pares'].append(numero)
    else:
      resultado['impares'].append(numero)

  return resultado

while True:
  print("\n--- Classificador de Pares e Impares ---")
  entrada = input("Digite os numeros separados por espaco (ou 'f' pra sair): ")

  if entrada.lower() == 'f':
    break

  try:
    listaConvertida = []
    for x in entrada.split():
      listaConvertida.append(int(x))

    if len(listaConvertida) == 0:
      print("A lista ta vazia, digite pelo menos um numero.")
      continue

    classificacao = classificar_pares_impares(listaConvertida)

    print(f"Numeros pares: {classificacao['pares']}")
    print(f"Numeros impares: {classificacao['impares']}")

  except ValueError:
    print("Formato incorreto, certifique q digitou apenas numeros inteiros separados por espaco.")
```

```python
# 46 Calcule a média e o desvio padrão de uma lista de números.

'''
a formula  do desvio padrão pede pra  pegar cada numero da lista,
subtrair da media, elevar esse resultado ao quadrado e somar tudo. dps a gente divide isso
pelo tamanho da lista (pra achar a variancia) e por fim tira a raiz quadrada disso.
importei o modulo math pra usar a funcao sqrt q ja tira a raiz quadrada direto.
usei o early return pra evitar divisao por zero caso chegue uma lista vazia na funcao.
'''
import math

def calcular_media_desvio(listaNumeros):
  tamanhoLista = len(listaNumeros)

  if tamanhoLista == 0:
    return 0, 0

  soma = sum(listaNumeros)
  media = soma / tamanhoLista

  somaDosQuadrados = 0
  for numero in listaNumeros:
    somaDosQuadrados += (numero - media) ** 2

  variancia = somaDosQuadrados / tamanhoLista
  desvioPadrao = math.sqrt(variancia)

  return media, desvioPadrao

while True:
  print("\n--- Media e Desvio Padrao ---")
  entrada = input("Digite os numeros separados por espaco (ou 'f' pra sair): ")

  if entrada.lower() == 'f':
    break

  try:

    listaValores = []
    for x in entrada.split():
      listaValores.append(float(x))

    if len(listaValores) == 0:
      print("A lista ta vazia, digite pelo menos um numero.")
      continue

    resultadoMedia, resultadoDesvio = calcular_media_desvio(listaValores)

    '''usei o :.2f pra formatar a saida e deixar so com duas casas decimais'''
    print(f"Media: {resultadoMedia:.2f}")
    print(f"Desvio Padrao: {resultadoDesvio:.2f}")

  except ValueError:
    print("Formato incorreto, certifique q digitou apenas numeros separados por espaco.")
```

```python
# 47 Conte quantas vezes cada número aparece em uma lista.

'''
pra contar a frequencia de cada numero eu usei um dicionario pq e a melhor
estrutura pra guardar a chave (o numero) e o valor (a quantidade de vezes q apareceu).
a logica e iterar sob a lista e ver se o numero ja existe no dicionario.
se ja existir, a gente soma 1 na contagem dele. se nao existir, a gente cria a chave dele ja com o valor 1.
no input do while, mantive a msm logica do exercicio anterior usando o laco for
normal com append pra ler os numeros do split
'''
def contar_frequencia(listaNumeros):
  frequencia = {}

  for numero in listaNumeros:
    if numero in frequencia:
      frequencia[numero] += 1
    else:
      frequencia[numero] = 1

  return frequencia

while True:
  print("\n--- Contador de Frequencia ---")
  entrada = input("Digite os numeros separados por espaco (ou 'f' pra sair): ")

  if entrada.lower() == 'f':
    break

  try:

    listaConvertida = []
    for x in entrada.split():
      listaConvertida.append(int(x))

    if len(listaConvertida) == 0:
      print("A lista ta vazia, digite pelo menos um numero.")
      continue

    resultadoFrequencia = contar_frequencia(listaConvertida)

    print("\nFrequencia de cada numero:")

    for chave, valor in resultadoFrequencia.items():
      print(f"- Numero {chave} apareceu {valor} vez(es)")

  except ValueError:
    print("Formato incorreto, certifique q digitou apenas numeros inteiros separados por espaco.")
```

```python
# 48 Implemente o algoritmo de ordenação por inserção (Insertion Sort).

'''
pra resolver essa questao eu fiz um laco for comecando do indice 1 e guardei o valor atual numa variavel chamada chave.
dps criei um while pra ir olhando os numeros q estao nos indices anteriores a essa chave.
a logica q eu fiz no while e: se o numero de tras for maior q a chave, eu copio ele uma posicao pra frente.
vou decrementando o indice j ate achar um numero menor q a chave ou ate chegar no comeco da lista (j < 0).
quando o while para, eu insiro a chave na posicao certa q ficou livre.
no input, mantive a leitura com for e append dentro do try, e usei o metodo copy() na chamada da funcao pra nao sobrescrever a lista original.
'''
def insertion_sort(lista):
  tamanhoLista = len(lista)

  for i in range(1, tamanhoLista):
    chave = lista[i]
    j = i - 1

    while j >= 0 and chave < lista[j]:
      lista[j + 1] = lista[j]
      j -= 1

    lista[j + 1] = chave

  return lista

while True:
  print("\n--- Ordenacao por Insercao (Insertion Sort) ---")
  entrada = input("Digite os numeros separados por espaco (ou 'f' pra sair): ")

  if entrada.lower() == 'f':
    break

  try:
    '''lendo os dados com for e append normal'''
    listaNumeros = []
    for x in entrada.split():
      listaNumeros.append(float(x))

    if len(listaNumeros) == 0:
      print("A lista ta vazia, digite pelo menos um numero.")
      continue

    '''usei o copy() pra mandar uma copia e nao alterar a original'''
    listaOrdenada = insertion_sort(listaNumeros.copy())

    print(f"Lista original: {listaNumeros}")
    print(f"Lista ordenada: {listaOrdenada}")

  except ValueError:
    print("Formato incorreto, certifique q digitou apenas numeros separados por espaco.")
```

```python
# 49 Calcule a soma dos dígitos de um número inteiro fornecido

'''
pra calcular a soma dos digitos eu usei a conversao pra string.
assim da pra usar um laco for normal pra iterar sob cada caractere do numero, bem parecido com oq a gente faz percorrendo um array ou charAt no java.
dentro do for, eu converto cada pedaco de volta pra int e vou guardando o valor na variavel soma.
tbm usei a funcao abs() no comeco pra pegar o valor absoluto, assim se o usuario digitar um numero negativo o sinal de menos e ignorado e nao quebra a logica.
no input mantive o nosso loop padrao com try except pra validar.
'''
def somar_digitos(numero):
  numeroLimpo = abs(numero)
  soma = 0

  for digito in str(numeroLimpo):
    soma += int(digito)

  return soma

while True:
  print("\n--- Soma dos Digitos ---")
  entrada = input("Digite um numero inteiro (ou 'f' pra sair): ")

  if entrada.lower() == 'f':
    break

  try:
    valor = int(entrada)

    resultadoSoma = somar_digitos(valor)
    print(f"A soma dos digitos de {valor} é {resultadoSoma}")

  except ValueError:
    print("Formato incorreto, vc precisa digitar um numero inteiro valido.")
```

```python
# 50 Calcule a frequência de cada letra em uma string fornecida pelo usuário.

'''
pra resolver essa questao eu criei um dicionario vazio e fiz um laco for pra iterar sob cada caractere da string.
dentro do for eu chamei a funcao isalpha() nativa do python pra garantir q o loop so vai processar oq for letra msm, ignorando espaco e pontuacao.
tbm joguei um lower() pra converter tudo pra minusculo, senao o codigo ia contar 'A' e 'a' como duas letras diferentes.
a logica do if e a msm do exercicio 47: verifico se a letra ja existe como chave no dicionario. se existir, somo 1 no valor. se nao, crio a chave iniciando com 1.
no bloco do while, coloquei um if com strip() pra barrar o usuario caso ele so de enter ou digite apenas espacos.
'''
def contar_frequencia_letras(texto):
  frequencia = {}

  for caractere in texto:
    if caractere.isalpha():
      letra = caractere.lower()

      if letra in frequencia:
        frequencia[letra] += 1
      else:
        frequencia[letra] = 1

  return frequencia

while True:
  print("\n--- Frequencia de Letras ---")
  entrada = input("Digite uma palavra ou frase (ou 'f' pra sair): ")

  if entrada.lower() == 'f':
    break

  if len(entrada.strip()) == 0:
    print("O texto ta vazio, digite alguma coisa.")
    continue

  resultado = contar_frequencia_letras(entrada)

  if len(resultado) == 0:
    print("Nao achei nenhuma letra no texto (so tinha numero ou pontuacao).")
  else:
    print("\nFrequencia de cada letra:")
    '''iterando sob o dicionario pra formatar a saida bonitinha no console'''
    for chave, valor in resultado.items():
      print(f"- Letra '{chave}': {valor} vez(es)")
```
