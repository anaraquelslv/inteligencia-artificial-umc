Crie dois arrays de números inteiros aleatórios de tamanho 20 e calcule o ângulo entre eles (produto escalar normalizado).

```python
import numpy as np

array_a = np.random.randint(1, 11, size=20)
array_b = np.random.randint(1, 11, size=20)

print("Array A:")
print(array_a)
print()

print("Array B:")
print(array_b)
print()

produto_escalar = np.dot(array_a, array_b)

norma_a = np.linalg.norm(array_a)
norma_b = np.linalg.norm(array_b)

cosseno = produto_escalar / (norma_a * norma_b)

cosseno_ajustado = np.clip(cosseno, -1.0, 1.0)

angulo_radianos = np.arccos(cosseno_ajustado)

angulo_graus = np.degrees(angulo_radianos)

print(f"Cosseno calculado: {cosseno_ajustado:.4f}")
print(f"O angulo entre os vetores e de {angulo_graus:.2f} graus")
```

Calcule a integral da função $f(x) = x^2 \sin(x)$ no intervalo de 0 a $\pi$.

```python
import numpy as np
from scipy import integrate

def funcao_f(x):
    return (x ** 2) * np.sin(x)

limite_inferior = 0
limite_superior = np.pi

resultado_completo = integrate.quad(funcao_f, limite_inferior, limite_superior)

valor_integral = resultado_completo[0]
erro_estimado = resultado_completo[1]

print("Resultado do calculo da integral:")
print(valor_integral)
print()

print("Erro estimado da operacao:")
print(erro_estimado)
```

Use a função de otimização de SciPy para maximizar $f(x) = -x^2 + 4x + 1$.

```python
import numpy as np
from scipy import optimize

def funcao_invertida(x):
    return -(-(x ** 2) + (4 * x) + 1)

chute_inicial = [0.0]

resultado = optimize.minimize(funcao_invertida, chute_inicial)

x_maximo = resultado.x[0]

valor_maximo_y = -resultado.fun

print("Coordenada X onde ocorre o ponto máximo:")
print(np.round(x_maximo, 4))
print()

print("Valor máximo alcançado pela função original:")
print(np.round(valor_maximo_y, 4))
```

Calcule a transformada de Fourier de um sinal composto de $\sin(2\pi t)$ e $\cos(4\pi t)$.

```python
import numpy as np
from scipy import fft

quantidade_amostras = 500
espacamento = 1.0 / quantidade_amostras

tempo = np.linspace(0.0, 1.0, quantidade_amostras, endpoint=False)

sinal_misto = np.sin(2 * np.pi * tempo) + np.cos(4 * np.pi * tempo)

resultado_fft_complexo = fft.fft(sinal_misto)

magnitude_frequencias = np.abs(resultado_fft_complexo)

regua_frequencias = fft.fftfreq(quantidade_amostras, espacamento)

frequencias_positivas = regua_frequencias[:quantidade_amostras//2]
magnitudes_positivas = magnitude_frequencias[:quantidade_amostras//2]

limiar_deteccao = 50
indices_picos = np.where(magnitudes_positivas > limiar_deteccao)

print("Frequências dominantes encontradas no sinal (em Hz):")
print(np.round(frequencias_positivas[indices_picos], 2))
```

Calcule a integral dupla da função $f(x,y) = xy + x^2$ no intervalo $[0, 2]$ para $x$ e $[0, 1]$ para $y$.

```python
from scipy import integrate

def funcao_f(y, x):
    return (x * y) + (x ** 2)

limite_x_inferior = 0
limite_x_superior = 2

limite_y_inferior = lambda x: 0
limite_y_superior = lambda x: 1

resultado_completo = integrate.dblquad(funcao_f, limite_x_inferior, limite_x_superior, limite_y_inferior, limite_y_superior)

valor_integral = resultado_completo[0]
erro_estimado = resultado_completo[1]

print("Resultado do cálculo da integral dupla:")
print(valor_integral)
print()

print("Erro estimado da operação:")
print(erro_estimado)
```

Utilize a função de otimização multivariada para minimizar $f(x, y) = (x-2)^2 + (y-3)^2$.

Resolva a equação diferencial de segunda ordem $d^2y/dx^2 + y = 0$, com condições iniciais $y(0) = 1$ e $y'(0) = 0$.

```python
import numpy as np
from scipy import integrate

def sistema_equacoes(x, u):
    y1 = u[0]
    y2 = u[1]

    dy1_dx = y2
    dy2_dx = -y1

    return [dy1_dx, dy2_dx]

intervalo = (0, 10)

condicoes_iniciais = [1.0, 0.0]

pontos_x = np.linspace(0, 10, 11)

resultado = integrate.solve_ivp(sistema_equacoes, intervalo, condicoes_iniciais, t_eval=pontos_x)

print("Valores de X avaliados:")
print(resultado.t)
print()

print("Resultados de Y calculados para cada X:")
print(np.round(resultado.y[0], 4))
```

Realize a transformada de Fourier de um sinal discreto composto de senos e cossenos.


```python
import numpy as np
from scipy import fft

quantidade_amostras = 400
taxa_amostragem = 1.0 / 100.0

tempo = np.linspace(0.0, quantidade_amostras * taxa_amostragem, quantidade_amostras, endpoint=False)

sinal_discreto = np.sin(2 * np.pi * 10 * tempo) + np.cos(2 * np.pi * 20 * tempo)

transformada = fft.fft(sinal_discreto)
frequencias = fft.fftfreq(quantidade_amostras, taxa_amostragem)

magnitudes = np.abs(transformada) * (2.0 / quantidade_amostras)

frequencias_positivas = frequencias[:quantidade_amostras//2]
magnitudes_positivas = magnitudes[:quantidade_amostras//2]

indices_picos = np.where(magnitudes_positivas > 0.5)

print("Frequências puras encontradas no sinal discreto (em Hz):")
print(np.round(frequencias_positivas[indices_picos], 2))
```

Agrupe o DataFrame pela coluna Categoria e calcule o total de produtos por categoria.

```python
import pandas as pd

dados_produtos = {
    'Produto': ['Teclado', 'Mouse', 'Monitor', 'Cadeira', 'Headset', 'Mousepad'],
    'Categoria': ['Periféricos', 'Periféricos', 'Hardware', 'Móveis', 'Áudio', 'Periféricos'],
    'Preço': [150.00, 80.00, 800.00, 1200.00, 250.00, 40.00],
    'Quantidade': [10, 25, 5, 3, 12, 30]
}

tabela = pd.DataFrame(dados_produtos)

total_por_categoria = tabela.groupby('Categoria')['Quantidade'].sum().reset_index()

print("Tabela original:")
print()
print(tabela)

print("\n-----------------------------------\n")

print("Total de produtos por categoria:")
print()
print(total_por_categoria)
```

Preencha os valores nulos de uma coluna numérica com a mediana dessa coluna.

```python
import pandas as pd
import numpy as np

dados_produtos = {
    'Produto': ['Teclado', 'Mouse', 'Monitor', 'Cadeira', 'Headset'],
    'Preço': [150.00, np.nan, 800.00, 1200.00, np.nan],
    'Quantidade': [10, 25, 5, 3, 12]
}

tabela = pd.DataFrame(dados_produtos)

mediana_preco = tabela['Preço'].median()

tabela['Preço'] = tabela['Preço'].fillna(mediana_preco)

print(f"Mediana calculada para a coluna Preço: R$ {mediana_preco:.2f}\n")
print("DataFrame com os valores nulos preenchidos:")
print()
print(tabela)
```

Verifique quais colunas de um DataFrame contêm valores duplicados.

```python
import pandas as pd

dados_produtos = {
    'ID_Produto': [101, 102, 103, 104, 105],
    'Categoria': ['Periféricos', 'Hardware', 'Periféricos', 'Móveis', 'Áudio'],
    'Disponivel': [True, True, False, True, False]
}

tabela = pd.DataFrame(dados_produtos)

colunas_com_duplicatas = []

for coluna in tabela.columns:
    if tabela[coluna].duplicated().any():
        colunas_com_duplicatas.append(coluna)

print("DataFrame analisado:")
print(tabela)
print()
print("Colunas que contêm valores duplicados:")
print(colunas_com_duplicatas)
```

Crie uma tabela dinâmica a partir do DataFrame, calculando a soma de Quantidade por Categoria.

```python
import pandas as pd

dados_produtos = {
    'Produto': ['Teclado', 'Mouse', 'Monitor', 'Cadeira', 'Headset', 'Mousepad'],
    'Categoria': ['Periféricos', 'Periféricos', 'Hardware', 'Móveis', 'Áudio', 'Periféricos'],
    'Quantidade': [10, 25, 5, 3, 12, 30]
}

tabela = pd.DataFrame(dados_produtos)

tabela_dinamica = pd.pivot_table(
    tabela,
    values='Quantidade',
    index='Categoria',
    aggfunc='sum'
)

print("Tabela Dinâmica:")
print()
print(tabela_dinamica)
```

Mescle dois DataFrames baseados em uma coluna comum chamada ID utilizando um join.

```python
import pandas as pd

dados_cadastros = {
    'ID': [101, 102, 103, 104],
    'Produto': ['Teclado', 'Mouse', 'Monitor', 'Headset']
}

dados_estoque = {
    'ID': [101, 102, 103, 104],
    'Quantidade': [15, 30, 5, 12]
}

tabela_cadastros = pd.DataFrame(dados_cadastros)
tabela_estoque = pd.DataFrame(dados_estoque)

tabela_mesclada = pd.merge(tabela_cadastros, tabela_estoque, on='ID')

print(tabela_mesclada)
```

Crie um DataFrame com dados faltantes e use a interpolação linear para preenchê-los.

```python
import pandas as pd
import numpy as np

dados = {
    'Dia': [1, 2, 3, 4, 5, 6, 7],
    'Valor': [10.0, np.nan, 30.0, np.nan, np.nan, 60.0, 70.0]
}

tabela = pd.DataFrame(dados)

print("Tabela original (com dados faltantes):")
print()
print(tabela)
print()
print("-----------------------------------")
print()

tabela['Valor'] = tabela['Valor'].interpolate(method='linear')

print("Tabela após a interpolação linear:")
print()
print(tabela)
```

Crie um DataFrame a partir de um dicionário de listas e calcule a média de cada coluna numérica.

```python
import pandas as pd

dados = {
    'Produto': ['Teclado', 'Mouse', 'Monitor', 'Cadeira', 'Headset'],
    'Preço': [150.00, 80.00, 800.00, 1200.00, 250.00],
    'Quantidade': [10, 25, 5, 3, 12]
}

tabela = pd.DataFrame(dados)

medias = tabela.mean(numeric_only=True)

print("DataFrame criado:")
print()
print(tabela)
print()
print("Média de cada coluna numérica:")
print()
print(medias)
```
